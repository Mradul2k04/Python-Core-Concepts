NumPy (Intermediate)
 An online platform tracks time spent (in minutes) by students on 6 courses over 14 days using a NumPy array.
 
Task:
Compute total time spent per course


Identify courses where average daily engagement exceeds a threshold


Rank courses based on engagement


```python
import numpy as np
```

```python
np.random.seed(50)
```

```python
track_time=np.random.randint(10,120,size=(6,14))
print("Time spend on course in 14 days  ",track_time)
```

```python
total_time=track_time.sum(axis=1)
```

```python
print("Time spend to complete one course",total_time)
```

```python
threshold=60
average_daily_engagment=np.mean(track_time,axis=1)
print("The Average time ",average_daily_engagment)
```

```python
exceeds_threshold_mask=average_daily_engagment>threshold
course_exceeding=np.where(exceeds_threshold_mask)[0]
print(course_exceeding)
```

```python
course_ranking=total_time.argsort()[::-1]
print("Ranking of courses from 0 to 5",course_ranking)
```
