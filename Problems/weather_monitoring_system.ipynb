NumPy (Intermediate)
 A weather station stores hourly temperature readings for 7 days in a NumPy array of shape (7, 24).
 
Task:
Find the daily maximum and minimum temperature


Identify the day with the largest temperature variation


Replace outliers beyond ±2 standard deviations with the mean temperature




```python
import numpy as np
```

```python
np.random.seed(46)
```

```python
temperature_range=np.random.randint(-30,50,size=(7,24))
print("Temperature reading for 7 day per hours",temperature_range)
```

```python
daily_max=temperature_range.max(axis=1)
daily_min=temperature_range.min(axis=1)
print("The max temperature day wise",daily_max)
print("The min temperature day wise",daily_min)
```

```python
daily_temperature_variation=daily_max-daily_min
print(daily_temperature_variation)
```

```python
largest_variation_day=np.argmax(daily_temperature_variation)
print(largest_variation_day+1)
```

```python
mean_temperature=temperature_range.mean()
std_temp=temperature_range.std()
```

```python
lower_bound=mean_temperature-(2*std_temp)
upper_bound=mean_temperature+(2*std_temp)
```

```python
cleaned_temp=temperature_range.copy()
cleaned_temp
```

```python
mask=(cleaned_temp<lower_bound)| (cleaned_temp>upper_bound)
```

```python
cleaned_temp[mask]=mean_temperature
```

```python
print("Original Temperature data",temperature_range)
print("Temperature data after removing outliers",cleaned_temp)
```
