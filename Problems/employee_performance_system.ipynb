NumPy (Intermediate)
 An HR system stores employee performance ratings (scale 1–5) for 100 employees over 4 quarters in a NumPy array.
 
Task:
Normalize the ratings using min-max normalization


Calculate the average rating per employee


Identify employees whose average rating is above the company mean




```python
import numpy as np
```

```python
np.random.seed(45)
```

```python
ratings=np.random.randint(1,6,size=(100,4))
```

```python
re_min=ratings.min()
re_max=ratings.max()
normalization_ratings=(ratings-re_min)/(re_max-re_min)
print(normalization_ratings)
```

```python
average_ratings_per_employee=normalization_ratings.mean(axis=1)
print(average_ratings_per_employee)
```

```python
company_mean=normalization_ratings.mean()
print(company_mean)
```

```python
top_employee=np.where(average_ratings_per_employee>company_mean)[0]
```

```python
print(top_employee)
```
