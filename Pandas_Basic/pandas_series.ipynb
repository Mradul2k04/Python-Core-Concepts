Pandas Series

A Pandas Series is like a column in a table.It is a 1-D array holding data of any type

```python
#Importing pandas
import numpy as np
import pandas as pd
```

```python
country=['India','Pakistan','USA','Nepal','Srilanka']

pd.Series(country)
```

```python
runs=[13,2,4,5,8,1,5]

pd.Series(runs)
```

```python
#custom index

marks=[67,88,52,100]

subjects=['Maths','English','Science','Compuer']

pd.Series(marks,index=subjects)
```

```python
#setting name
marks=pd.Series(marks,index=subjects,name='Mradul marks')
marks
```

Series from Dict

```python
marks={
    'maths':65,
    'english':88,
    'hindi':58,
    'computer':100
}
marks_series=pd.Series(marks,name="Mradul marks")
marks_series
```

Series Attributes

```python
#size
marks_series.size
```

```python
#dtype
marks_series.dtype
```

```python
#name
marks_series.name
```

```python
#is_unique
marks_series.is_unique
```

```python
#index
marks_series.index
```

```python
#values
marks_series.values
```
